Hồi quy tuyến tính

Mỗi là một vector đặc trưng (đầu vào) trong không gian n chiều: 
xi= (xi1,xi2,...)^T
Mỗi chiều là một thuộc tính/đặc trưng của dữ liệu.
Trong đó W0 đôi khi được gọi là bias (độ lệch) – là hệ số không nhân với bất kỳ đặc trưng nào, chỉ là giá trị cộng thêm.


Với mỗi điểm dữ liệu x, mô hình dự đoán một giá trị 𝑦𝑥
Đối với dữ liệu tương lai z, mô hình dùng các trọng số đã học để tính toán đầu ra dự đoán 𝑓(𝑧)
Mục tiêu: dự đoán càng gần giá trị thật càng tốt.


Vì có vô hạn hàm có thể mô tả mối quan hệ giữa x và y, ta cần một loss function để chọn ra hàm tốt nhất — là hàm có khả năng dự đoán chính xác nhất cho dữ liệu tương lai.


Vì chúng ta không biết phân phối thực của dữ liệu đầu vào 
𝑥 nên không thể tính chính xác kỳ vọng Ex[r(x)]. Thay vào đó, ta ước lượng nó bằng trung bình trên tập huấn luyện.
